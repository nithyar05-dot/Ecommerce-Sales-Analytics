```python
import pandas as pd

df = pd.read_csv("ecommerce_sales_week4.csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nRows Containing Missing Values:")
display(df[df.isnull().any(axis=1)])
```

```python
# Fill missing Quantity with its median
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())

# Calculate missing Sales using Quantity × Unit_Price
df["Sales"] = df["Sales"].fillna(
    df["Quantity"] * df["Unit_Price"]
)

# Fill missing Customer_Rating with its median
df["Customer_Rating"] = df["Customer_Rating"].fillna(
    df["Customer_Rating"].median()
)

# Verify missing values are handled
print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Count duplicate rows before removal
print("Duplicates Before Removal:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Verify duplicates have been removed
print("Duplicates After Removal:", df.duplicated().sum())

# Check the updated dataset size
print("Final Dataset Shape:", df.shape)
```

```python
# Verify the cleaned dataset
print("Final Dataset Shape:", df.shape)
print("Total Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Rows:", df.duplicated().sum())

# Save the cleaned dataset
df.to_csv("ecommerce_sales_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

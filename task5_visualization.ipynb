```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ecommerce_sales_cleaned.csv")

print("Dataset loaded:", df.shape)
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(
    ascending=False
)

category_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("chart1_sales_by_category.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values(
    ascending=False
)

region_profit.plot(kind="bar", figsize=(8, 5))

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("chart2_profit_by_region.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=10, edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales Amount")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.savefig("chart3_dist_sales",dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Discount_Percent"], df["Sales"])

plt.title("Discount Percentage vs Sales")
plt.xlabel("Discount Percentage")
plt.ylabel("Sales Amount")
plt.tight_layout()
plt.savefig("chart4_discount%_vs_sales",dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Customer_Rating"], bins=8, edgecolor="black")

plt.title("Distribution of Customer Ratings")
plt.xlabel("Customer Rating")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.savefig("chart5_dist_customer_rating",dpi=300, bbox_inches="tight")
plt.show()
```

```python
product_profit = df.groupby("Product")["Profit"].sum().sort_values(
    ascending=False
)

product_profit.plot(kind="bar", figsize=(10, 5))

plt.title("Total Profit by Product")
plt.xlabel("Product")
plt.ylabel("Total Profit")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("chart6_profit_by_product",dpi=300, bbox_inches="tight")
plt.show()
```
